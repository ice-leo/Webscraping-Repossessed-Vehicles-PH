## Extract Dataframes

```python
locations = ['bacolod', 'cagayan_de_oro', 'cebu', 'dagupan', 'davao', 'guiguinto', 'iloilo', 'lipa', 'naga', 'san_fernando']
#Read PDF File
for loc in locations:
  pdf_path = f'Desktop/Webscraping Datasets (BPI)/{loc}.pdf'
  dfs = tabula.io.read_pdf(pdf_path, pages='all', pandas_options={"header": None}, lattice=True)

  df = pd.concat(dfs, ignore_index=True)

  #Remove Unnecessary Column
  df = df.iloc[2:]
  df.columns = ["unit_description", "kilometrage", "pn/cs", "color","reg_bidding_price", "min_bidding_price"]

  #Remove Headers
  df = df[~((df['unit_description'] == 'UNIT DESCRIPTION') | (df['unit_description'].isna()))]

  # Reset the index if needed
  df.reset_index(drop=True, inplace=True)

  #Add Location Column
  df['location'] = loc

  #Add Year Column
  df['year'] = df['unit_description'].str[:4].astype(int)

  #Add Transmission Column
  def extract_transmission(unit_description):
      if 'A/T' in unit_description:
          return 'A/T'
      elif 'M/T' in unit_description:
          return 'M/T'
      else:
          return np.nan

  # Apply the function to create the 'transmission_type' column
  df['transmission_type'] = df['unit_description'].apply(extract_transmission)

  # Filter out the rows where 'unit_description' is 'UNIT DESCRIPTION' or NaN
  df = df[~((df['unit_description'] == 'UNIT DESCRIPTION') | (df['unit_description'].isna()))]

  # Reset the index if needed
  df.reset_index(drop=True, inplace=True)

  #Name the dataframe
  exec('df_{} = df'.format(loc))
```

```python
import numpy as np
from bs4 import BeautifulSoup
from tqdm import tqdm
import json
import requests
import re
import time
import pandas as pd
```

## Parse HTML of all Pages

```python
#Enter number of pages
num_pages = 359
```

```python
#Pages List
pages = ['p{}'.format(i) for i in range(1, num_pages + 1)]

#Get the html of all pages
for page in pages:
  url = r"https://philkotse.com/used-cars-for-sale/{}".format(page)
  content_html = requests.get(url)
  content = content_html.content
  exec('{}_soup = BeautifulSoup(content, "html.parser")'.format(page))
  print("Completed: " + str(pages.index(page) + 1))
print("Complete")
```

## Get the links of all cars

```python
'''
start = <h3 class="title"><a href=
end = </a></h3>
'''
```

```python
links = []

for page in pages:
  exec('''num_links = len(str(p{}_soup).split('<h3 class="title"><a href="')) - 1'''.format(str(pages.index(page) + 1)))
  for i in range(1, num_links + 1):
    links.append(str(eval(page+'_soup')).split('<h3 class="title"><a href="')[i].split('">')[0])
    print("Links added: " + str(len(links)))

print("Complete")
```

```python
from collections import Counter

# Count occurrences of each element
count = Counter(links)

# Find elements with duplicates
duplicates = {item: freq for item, freq in count.items() if freq > 1}

# Number of duplicates
num_duplicates = sum(freq - 1 for freq in duplicates.values())

print("Duplicates:", duplicates)
print("Number of duplicates:", num_duplicates)
```

```python
#Remove Null Values
df = df.dropna()

#Convert Year to Integer
df['Year'] = df['Year'].astype(int)

#Convert Price to Integer
df['Price'] = df['Price'].replace('[^0-9]', '', regex=True).astype(int)

## Clean 'Kilometrage' values and convert to 'Mileage' in miles as an integer
df['Kilometrage'] = df['Kilometrage'].str.replace('[^0-9]', '', regex=True)
df['Kilometrage'] = pd.to_numeric(df['Kilometrage'], errors='coerce')  # Convert to numeric, set errors to NaN
df = df.dropna(subset=['Kilometrage'])  # Drop rows where 'Kilometrage' could not be converted
df['Mileage'] = (df['Kilometrage'] * 0.621371).astype(int)

#Remove Excellent\r Transmission
df = df[df['Transmission'] != 'Excellent\r']

#Convert 'Automatic\r', 'Automatic \r' Transmissions to 'Automatic'
df['Transmission'] = df['Transmission'].str.replace('Automatic\r', 'Automatic').str.replace('Automatic \r', 'Automatic')
```
